```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("/content/adult (1).csv")

# Inspect the dataset
print("Dataset shape:", df.shape)

print("\nDataset information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())

print("\nMissing values BEFORE replacing '?':")
print(df.isnull().sum())
```

```python
# Replace " ?" with NaN
df.replace(" ?", np.nan, inplace=True)

# Check missing values again
print("Missing values AFTER replacing '?':")
print(df.isnull().sum())
```

```python
# Calculate missing value count
missing_count = df.isnull().sum()

# Calculate missing value percentage
missing_percentage = (df.isnull().mean() * 100)

# Combine into a DataFrame
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

# Sort by missing count descending
missing_summary = missing_summary.sort_values(
    by="Missing Count",
    ascending=False
)

display(missing_summary)
```

```python
# Fill missing categorical values with "Unknown"
df["workclass"] = df["workclass"].fillna("Unknown")
df["occupation"] = df["occupation"].fillna("Unknown")

# Also handle native-country
df["native.country"] = df["native.country"].fillna("Unknown")

# Check remaining missing values
print("Missing values after categorical imputation:")
print(df.isnull().sum())
```

```python
# Find numeric columns with missing values
numeric_missing = df.select_dtypes(include=np.number).isnull().sum()

print("Missing values in numeric columns:")
print(numeric_missing)
```

```python
# Impute missing numeric values with the median
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:")
print(df.isnull().sum())
```

```python
# Count exact duplicate rows
exact_duplicates = df.duplicated().sum()

print("Number of exact duplicate rows:", exact_duplicates)

# Remove exact duplicate rows
df.drop_duplicates(inplace=True)

print("Shape after removing exact duplicates:", df.shape)
```

```python
# Check column names
print(df.columns.tolist())
```

```python
# Remove the target column for duplicate checking
features_only = df.drop(columns=["income"])

# Count duplicate rows based only on features
feature_duplicates = features_only.duplicated().sum()

print("Duplicates ignoring income/target column:", feature_duplicates)
```

```python
print("Unique education values:")
print(df["education"].unique())

print("\nUnique marital-status values:")
print(df["marital.status"].unique())

print("\nUnique native-country values:")
print(df["native.country"].unique())
```

```python
# Strip leading/trailing whitespace from all object columns
object_columns = df.select_dtypes(include="object").columns

for col in object_columns:
    df[col] = df[col].str.strip()

# Check unique values again
print("Education:")
print(df["education"].unique())

print("\nMarital Status:")
print(df["marital.status"].unique())

print("\nNative Country:")
print(df["native.country"].unique())
```

```python
# Descriptive statistics for numeric columns
display(df.describe())
```

```python
# Descriptive statistics for categorical columns
display(df.describe(include="object"))
```

```python
# Calculate mean and median
age_mean = df["age"].mean()
age_median = df["age"].median()

hours_mean = df["hours.per.week"].mean()
hours_median = df["hours.per.week"].median()

print("Age:")
print("Mean:", age_mean)
print("Median:", age_median)

print("\nHours-per-week:")
print("Mean:", hours_mean)
print("Median:", hours_median)
```

```python
age_difference = abs(age_mean - age_median)
hours_difference = abs(hours_mean - hours_median)

print(f"Difference between age mean and median: {age_difference:.2f}")
print(f"Difference between hours-per-week mean and median: {hours_difference:.2f}")

if age_difference < 2:
    print("Age: Mean and median are relatively close.")
else:
    print("Age: Mean and median differ noticeably.")

if hours_difference < 2:
    print("Hours-per-week: Mean and median are relatively close.")
else:
    print("Hours-per-week: Mean and median differ noticeably.")
```

```python
# Frequency of occupations
occupation_counts = df["occupation"].value_counts()

print("Occupation frequencies:")
display(occupation_counts)

print("\nMost common occupation:")
print(occupation_counts.idxmax())

print("Number of people:")
print(occupation_counts.max())
```

```python
# Percentage distribution of sex
sex_percentage = df["sex"].value_counts(normalize=True) * 100

print("Sex percentage distribution:")
display(sex_percentage.round(2))
```

```python
# Percentage distribution of income
income_percentage = df["income"].value_counts(normalize=True) * 100

print("Income percentage distribution:")
display(income_percentage.round(2))
```

```python
# Determine whether income classes are balanced
income_counts = df["income"].value_counts()

print("Income counts:")
display(income_counts)

ratio = income_counts.min() / income_counts.max()

print(f"Minority-to-majority class ratio: {ratio:.2f}")

if ratio >= 0.8:
    print("The income classes are relatively balanced.")
else:
    print("The income classes are imbalanced.")
```

```python
# Check which education-num values correspond to each education level
education_mapping = df.groupby("education")["education.num"].unique()

display(education_mapping)
```

```python
# Find education levels that have more than one education-num value
inconsistent_education = education_mapping[
    education_mapping.apply(len) > 1
]

print("Inconsistent education mappings:")
display(inconsistent_education)

if len(inconsistent_education) == 0:
    print("No inconsistencies found.")
else:
    print("Inconsistencies were found.")
```

```python
import sqlite3

# Read the RAW CSV again so SQL works on the original data
raw_df = pd.read_csv("/content/adult (1).csv")

# Create an in-memory SQLite database
connection = sqlite3.connect(":memory:")

# Store CSV data in SQLite table
raw_df.to_sql(
    "adult_income",
    connection,
    if_exists="replace",
    index=False
)

print("CSV successfully loaded into SQLite table: adult_income")
```

```python
# Display first 5 rows from SQLite
sql_check = pd.read_sql_query(
    "SELECT * FROM adult_income LIMIT 5",
    connection
)

display(sql_check)
```

```python
# SQL query
query = """
SELECT *
FROM adult_income
WHERE age > 30
"""

# Execute SQL query and load result into pandas
sql_df = pd.read_sql_query(query, connection)

print("SQL extraction shape:", sql_df.shape)

display(sql_df.head())
```

```python
# Number of rows in original CSV satisfying age > 30
expected_rows = (raw_df["age"] > 30).sum()

print("Expected number of rows where age > 30:", expected_rows)
print("Rows returned by SQL:", len(sql_df))

# Compare columns
print("\nOriginal CSV columns:", len(raw_df.columns))
print("SQL result columns:", len(sql_df.columns))

if len(sql_df) == expected_rows:
    print("\nSQL extraction is correct.")
else:
    print("\nSQL extraction does not match the expected result.")
```

```python
# Pandas equivalent of the SQL query
pandas_df = raw_df[raw_df["age"] > 30].copy()

print("Pandas filtering shape:", pandas_df.shape)
print("SQL filtering shape:", sql_df.shape)

if pandas_df.shape == sql_df.shape:
    print("\nSQL and pandas extraction have the same shape.")
else:
    print("\nSQL and pandas extraction have different shapes.")
```

## Task 10 — Data Profiling Summary Report

### Dataset Overview

The dataset consists of **32,561 records and 15 columns** in its original form. After removing exact duplicate records, the cleaned dataset contains **32,537 rows and 15 columns**. This is the **UCI Adult Income dataset**, which contains information about individuals from the 1994 U.S. Census. The variables describe characteristics such as age, education, employment, marital status, occupation, and working hours, with the income category indicating whether an individual's income is **<=50K or >50K**.

### Data Quality Issues Found

Several data quality problems were identified during the profiling process:

- **Missing data:** The dataset originally stored missing entries as `"?"`, which were not recognized as missing by pandas. After converting these values to `NaN`, **occupation had 1,843 missing values, workclass had 1,836, and native-country had 583**.
- **Duplicate records:** There were **24 exact duplicate rows**. When the income column was excluded from the duplicate check, **25 duplicate matches** were identified. The additional matching records had different income labels, so they were retained because the same recorded characteristics can occur for different individuals.
- **Categorical inconsistencies:** Several categorical fields contained unnecessary whitespace around their values. This could cause identical categories to be interpreted as separate values, so whitespace was removed from the categorical columns.

### Key Findings

The analysis showed that the income variable is noticeably **imbalanced**. Approximately **75.9%** of observations belong to the `<=50K` group, while only about **24.1%** belong to the `>50K` group. The sex variable is also unevenly distributed, with approximately **66.9% Male** and **33.1% Female**.

For occupation, **Prof-specialty** is the most frequently occurring category, with **4,136 observations**. Craft-repair and Exec-managerial are also among the more common occupations.

The average age is approximately **38.6 years**, compared with a median of **37 years**. Similarly, the average number of hours worked per week is around **40.4 hours**, while the median is **40 hours**. Since the mean and median values are fairly close for both variables, there does not appear to be a major difference between the center of these distributions, although some mild right skew may be present.

The relationship between `education` and `education-num` was also checked. **No inconsistencies were found**: every education category corresponds to one specific education-num value.

### Cleaning Decisions

The `"?"` placeholders were first converted into proper `NaN` values so that missing data could be correctly identified and handled. Missing values in **workclass** and **occupation** were replaced with an `"Unknown"` category instead of deleting those observations, which preserves the available information in the remaining columns.

For **native-country**, the missing entries were filled using the **mode, United-States**, because the missing proportion was relatively small at approximately **1.8%** and the column is strongly dominated by that country.

The **24 exact duplicate rows** were removed from the cleaned dataset. However, records that were identical in their attributes but differed in the income label were not removed because their different outcomes may represent valid observations.

Finally, leading and trailing spaces were removed from categorical variables to ensure that the same category was not accidentally stored under multiple slightly different strings.
